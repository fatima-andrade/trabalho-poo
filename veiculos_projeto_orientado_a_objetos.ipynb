```python
%%writefile veiculo.py
class Veiculo:
    def __init__(self, marca, modelo, ano, placa):
        self.marca = marca
        self.modelo = modelo
        self.ano = ano
        self.placa = placa

    def exibir_informacoes(self):
        print(f"Marca: {self.marca}")
        print(f"Modelo: {self.modelo}")
        print(f"Ano: {self.ano}")
        print(f"Placa: {self.placa}")

    def acelerar(self):
        print(f"O {self.modelo} está acelerando...")

    def frear(self):
        print(f"O {self.modelo} está freando...")
```

```python
%%writefile carro.py
from veiculo import Veiculo

class Carro(Veiculo):
    def __init__(self, marca, modelo, ano, placa, numero_portas):
        super().__init__(marca, modelo, ano, placa)
        self.numero_portas = numero_portas

    def acelerar(self):
        print(f"O carro {self.modelo} está acelerando com o motor a combustão.")

    def abrir_porta(self):
        print(f"A porta do carro {self.modelo} foi aberta.")
```

```python
%%writefile moto.py
from veiculo import Veiculo

class Moto(Veiculo):
    def __init__(self, marca, modelo, ano, placa, tipo):
        super().__init__(marca, modelo, ano, placa)
        self.tipo = tipo

    def acelerar(self):
        print(f"A moto {self.modelo} está acelerando com mais agilidade!")

    def empinar(self):
        print(f"A moto {self.modelo} está empinando!")
```

```python
%%writefile main.py
from carro import Carro
from moto import Moto

def menu():
    carro1 = Carro("Toyota", "Corolla", 2020, "ABC-1234", 4)
    moto1 = Moto("Honda", "CB500", 2021, "XYZ-5678", "Esportiva")

    while True:
        print("\n---- Menu ----")
        print("1. Exibir informações do carro")
        print("2. Exibir informações da moto")
        print("3. Acelerar carro")
        print("4. Acelerar moto")
        print("5. Sair")

        opcao = input("Escolha uma opção: ")

        if opcao == "1":
            carro1.exibir_informacoes()
        elif opcao == "2":
            moto1.exibir_informacoes()
        elif opcao == "3":
            carro1.acelerar()
        elif opcao == "4":
            moto1.acelerar()
        elif opcao == "5":
            print("Saindo...")
            break
        else:
            print("Opção inválida!")

menu()
```
